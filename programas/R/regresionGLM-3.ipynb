```r
# --- Script: Regresión Logística en Salud (R) ---

# 1. Instalación y Carga de Datos
if(!require(faraway)) install.packages("faraway")
library(faraway)
data(pima)

# 2. Análisis Exploratorio de Datos (EDA)
# Verificación de estructura y limpieza de valores nulos (NA)
pima_clean <- na.omit(pima)
summary(pima_clean[, c("bmi", "age", "test")])

# 3. Ajuste del Modelo GLM
# 'test' es la respuesta (0 = neg, 1 = pos). 
# Se especifica family = binomial para activar el enlace logit.
modelo_diabetes <- glm(test ~ bmi + age, 
                       data = pima_clean, 
                       family = binomial(link = "logit")) # [26, 27]

# 4. Inferencia y Evaluación de Resultados
summary(modelo_diabetes) # [28]

# Cálculo de los Odds Ratios (OR) con Intervalos de Confianza del 95%
OR_diabetes <- exp(cbind(OR = coef(modelo_diabetes), confint(modelo_diabetes))) # [18, 29]
print("Resultados de Odds Ratios y CIs:")
print(OR_diabetes)

# 5. Visualización del Modelo
# Probabilidad predicha frente al IMC
plot(pima_clean$bmi, modelo_diabetes$fitted.values, 
     col = "darkred", pch = 16, alpha = 0.5,
     main = "Probabilidad de Diabetes vs IMC",
     xlab = "IMC", ylab = "P(Diabetes)") # [7, 30]
```
