### Ejemplo 1: Predicción de precios de viviendas (Boston Housing)
Objetivo

Predecir el valor medio de las viviendas (medv) en función de variables socioeconómicas y estructurales.

```r
library(MASS)
library(tidyverse)
library(corrplot)
library(GGally)

data("Boston")

glimpse(Boston)
summary(Boston)

colSums(is.na(Boston))


ggplot(Boston, aes(medv)) +
  geom_histogram(bins = 30) +
  theme_minimal() +
  labs(title="Distribución Precio Vivienda")


correlation <- cor(Boston)

corrplot(
  correlation,
  method = "color",
  type = "upper"
)

ggpairs(
  Boston[,c("medv","rm","lstat","ptratio")]
)
```
