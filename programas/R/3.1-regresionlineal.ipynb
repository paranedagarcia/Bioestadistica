### Ejemplo 1: Regresión Lineal Simple - Factores Maternos y Peso al Nacer

En este ejemplo, utilizaremos el dataset birthwt (del paquete MASS), el cual contiene datos de 189 nacimientos en el Baystate Medical Center. El objetivo es determinar si el peso de la madre antes del embarazo (lwt) es un predictor lineal del peso del recién nacido (bwt).

### Análisis Exploratorio de Datos (EDA)

Antes de modelar, debemos comprender la distribución y la relación bivariada.

```r
library(MASS)
data("birthwt")

# Resumen estadístico inicial
summary(birthwt[, c("lwt", "bwt")])
```

```r
# Visualización de la relación
plot(birthwt$lwt, birthwt$bwt, 
     main="Relación Peso Materno vs Peso del Neonato",
     xlab="Peso Materno (libras)", ylab="Peso al Nacer (gramos)",
     pch=19, col="steelblue")
abline(lm(bwt ~ lwt, data=birthwt), col="red", lwd=2) # Línea de tendencia preliminar
```

```r
#Modelamiento
#Establecemos el modelo por el método de Mínimos Cuadrados Ordinarios (OLS).
model <- lm(bwt ~ lwt, data = birthwt)
summary(model)
```

```r
# Plot data with fitted regression line
plot(birthwt$lwt, birthwt$bwt, 
     main="Birth Weight vs Mother's Weight with Fitted Line",
     xlab="Mother's Weight (lbs)",
     ylab="Birth Weight (g)")
abline(model, col="red", lwd=2)
```

Interpretación: En el output de R, el coeficiente de lwt indica cuántos gramos aumenta el peso del neonato por cada libra adicional de peso materno. Es crítico observar el valor p asociado para rechazar la hipótesis nula ($H_0: \beta_1 = 0$).
