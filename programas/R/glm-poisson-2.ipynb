```r
# Parámetro de la distribución
lambda <- 10

# Probabilidades solicitadas
p3  <- dpois(3, lambda)
p15 <- dpois(15, lambda)
p4  <- dpois(4, lambda)

cat("P(X=3)  =", p3, "\n")
cat("P(X=15) =", p15, "\n")
cat("P(X=4)  =", p4, "\n")
```

```r
x <- 0:25
prob <- dpois(x, lambda)

barplot(prob,
        names.arg = x,
        col = "lightblue",
        main = "Distribución de Poisson (λ = 10)",
        xlab = "Número de pacientes",
        ylab = "Probabilidad")
```
