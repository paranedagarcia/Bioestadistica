### Rúbrica de Evaluación de 10 Pasos (ANOVA y EDA en R)

Los estudiantes deberán entregar un R Notebook (Rmd) que contenga el código ejecutable y la interpretación técnica de cada fase.

#### Fase I: Análisis Exploratorio de Datos (EDA)

1. **Configuración y Carga de Datos:**
* Cargar librerías esenciales (ej. `ggplot2`, `car`, `dplyr`).
* Importar el dataset y asegurar que la variable "X" sea un factor con dos o tres niveles.


2. **Estadística Descriptiva:**
* Generar una tabla con la media, desviación estándar y mediana  por cada grupo de tratamiento.
* Evaluar si el diseño es balanceado (mismo $n$ por grupo).


3. **Inspección Visual (Distribución):**
* Construir *boxplots* para identificar valores atípicos (outliers) y visualizar la dispersión.
* Utilizar gráficos de densidad para observar la simetría de las distribuciones.



#### Fase II: Verificación de Supuestos Estadísticos

4. **Prueba de Normalidad:**
* Ejecutar el test de Shapiro-Wilk para los residuos del modelo.
* Interpretar gráficos Q-Q (Quantile-Quantile) para confirmar visualmente la normalidad.


5. **Prueba de Homocedasticidad:**
* Aplicar el Test de Levene para evaluar la igualdad de varianzas entre protocolos ($H_0: \sigma_1^2 = \sigma_2^2 = \sigma_3^2$).


6. **Justificación Metodológica:**
* El alumno debe decidir explícitamente entre un **ANOVA de Fisher** (varianzas iguales) o un **ANOVA de Welch** (varianzas desiguales).



#### Fase III: Ejecución del Modelo e Inferencia

7. **Ajuste del Modelo:**
* Ejecutar el modelo adecuado utilizando `aov()` o `oneway.test()`.
* Formular formalmente la hipótesis nula ($H_0$) y la alternativa ($H_1$).


8. **Interpretación de la Tabla ANOVA:**
* **Valor F:** Explicar la razón entre la varianza entre grupos y la varianza intra-grupos.
* **Valor p:** Determinar la significancia estadística basada en un $\alpha = 0.05$.



#### Fase IV: Análisis Post-hoc y Reporte

9. **Comparaciones Múltiples (Post-hoc):**
* Si el ANOVA es significativo, realizar un test de Tukey HSD (homocedasticidad) o Games-Howell (heterocedasticidad).
* Interpretar los valores p ajustados para identificar qué protocolos difieren entre sí.


10. **Conclusión Clínica y Comunicación:**
* Traducir los hallazgos estadísticos a una recomendación clínica.
* Reportar el resultado bajo estándares científicos (ej. "Se encontró un efecto significativo del protocolo sobre los niveles de PCR, $F(2, 87) = X.XX, p < 0.05$").

### Criterios de Calificación

| Criterio | Excelente (100%) | Suficiente (70%) | Insuficiente (0%) |
| --- | --- | --- | --- |
| **Calidad del EDA** | Análisis visual y numérico exhaustivo; manejo de outliers. | Visualización mínima; resúmenes básicos. | Ausencia de análisis descriptivo. |
| **Validación de Supuestos** | Pruebas rigurosas de normalidad y varianza; elección correcta del modelo. | Realizó pruebas pero eligió el modelo incorrecto. | Ignoró los supuestos estadísticos. |
| **Rigor Estadístico** | Interpretación precisa del **Valor F** y **Valor p**. | Identificó significancia pero no explicó el estadístico F. | Interpretación errónea de los resultados. |
| **Traducción Clínica** | Recomendación clara, concisa y basada estrictamente en los datos. | Conclusión vaga o excesivamente técnica. | Sin conexión con el contexto clínico. |
